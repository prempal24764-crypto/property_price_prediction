## 📌 Objective

The objective of this project is to predict house prices based on various features like area, number of bedrooms, age, and location score using Machine Learning algorithms.

```python
import pandas as pd
import numpy as np

np.random.seed(42)

data = {
    'area': np.random.randint(500, 3000, 100),
    'bedrooms': np.random.randint(1, 5, 100),
    'age': np.random.randint(1, 20, 100),
    'location_score': np.random.randint(1, 10, 100)
}

df = pd.DataFrame(data)

# price generate (realistic logic)
df['price'] = (df['area'] * 0.05 +
               df['bedrooms'] * 10 +
               df['location_score'] * 15 -
               df['age'] * 2 +
               np.random.randint(0, 10, 100))

df.head()
```

## 📊 Data Description

- Area: Size of the house  
- Bedrooms: Number of bedrooms  
- Age: Age of the property  
- Location Score: Location quality  
- Price: Target variable

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor

X = df[['area', 'bedrooms', 'age', 'location_score']]
y = df['price']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Model 1
lr = LinearRegression()
lr.fit(X_train, y_train)

# Model 2 (Advanced 🔥)
rf = RandomForestRegressor()
rf.fit(X_train, y_train)
```

print("Linear Regression Accuracy:", lr.score(X_test, y_test))
print("Random Forest Accuracy:", rf.score(X_test, y_test))

```python
lr_pred = lr.predict(X_test)
rf_pred = rf.predict(X_test)

print("Linear Regression Accuracy:", lr.score(X_test, y_test))
print("Random Forest Accuracy:", rf.score(X_test, y_test))
```

```python
plt.scatter(y_test, rf_pred)
plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.title("Random Forest Prediction")
plt.show()
```
